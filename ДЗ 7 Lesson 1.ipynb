```python
from datetime import datetime
from typing import List, Optional, Dict


class CourtCase:
    def __init__(self, case_number: str):
        """
        Инициализация судебного дела.

        :param case_number: Номер дела (обязательный параметр)
        """
        self.case_number = case_number
        self.case_participants: List[str] = []
        self.listening_datetimes: List[Dict] = []
        self.is_finished: bool = False
        self.verdict: str = ""

    def set_a_listening_datetime(self, datetime_obj: datetime, description: str = "") -> None:
        """
        Добавляет судебное заседание в список listening_datetimes.

        :param datetime_obj: Дата и время заседания
        :param description: Описание заседания (опционально)
        """
        listening = {
            "datetime": datetime_obj,
            "description": description
        }
        self.listening_datetimes.append(listening)

    def add_participant(self, inn: str) -> None:
        """
        Добавляет участника в список case_participants.

        :param inn: ИНН участника
        """
        if inn not in self.case_participants:
            self.case_participants.append(inn)

    def remove_participant(self, inn: str) -> None:
        """
        Убирает участника из списка case_participants.

        :param inn: ИНН участника для удаления
        """
        if inn in self.case_participants:
            self.case_participants.remove(inn)

    def make_a_decision(self, verdict_text: str) -> None:
        """
        Выносит решение по делу.

        :param verdict_text: Текст решения суда
        """
        self.verdict = verdict_text
        self.is_finished = True

    def __str__(self) -> str:
        """Строковое представление дела."""
        status = "Завершено" if self.is_finished else "В процессе"
        return f"Дело №{self.case_number} — {status}"

    def __repr__(self) -> str:
        """Официальное представление объекта."""
        return f"CourtCase(case_number='{self.case_number}', is_finished={self.is_finished})"


# Пример использования:
if __name__ == "__main__":
    # Создание дела
    case = CourtCase("А40-12345/2026")

    # Добавление участников
    case.add_participant("7701234567")
    case.add_participant("7709876543")

    # Добавление судебных заседаний
    case.set_a_listening_datetime(datetime(2026, 9, 15, 10, 0), "Предварительное заседание")
    case.set_a_listening_datetime(datetime(2026, 9, 22, 14, 30), "Основное заседание")

    # Удаление участника
    case.remove_participant("7709876543")

    # Вынесение решения
    case.make_a_decision("Иск удовлетворить полностью. Ответчик обязан выплатить 500 000 рублей.")

    # Вывод информации
    print(case)  # Дело №А40-12345/2026 — Завершено
    print(f"Участники: {case.case_participants}")
    print(f"Заседания: {len(case.listening_datetimes)}")
    print(f"Решение: {case.verdict}")
```
