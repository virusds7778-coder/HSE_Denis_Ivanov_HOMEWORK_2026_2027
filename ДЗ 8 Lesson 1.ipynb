```python
class Contract:
    """Абстракция юридического договора."""

    def __init__(self, contract_number, party_a, party_b, subject, date):
        self.contract_number = contract_number
        self.party_a = party_a
        self.party_b = party_b
        self.subject = subject
        self.date = date
        self.is_signed = False
        self.is_active = True
        self.deadlines = []
        self.obligations = []
        self.signatures = []
        self.status = "действует"

    def sign(self, signer_name):
        """Подписать договор."""
        self.signatures.append(signer_name)
        self.is_signed = True
        print(f"Договор №{self.contract_number} подписан: {signer_name}")

    def add_deadline(self, deadline, description):
        """Добавить срок/дедлайн."""
        self.deadlines.append({"date": deadline, "description": description})
        print(f"Добавлен срок: {description} — {deadline}")

    def add_obligation(self, party, obligation_text):
        """Добавить обязательство для стороны."""
        self.obligations.append({"party": party, "text": obligation_text})
        print(f"Обязательство для {party}: {obligation_text}")

    def fulfill_obligation(self, obligation_index):
        """Выполнить обязательство."""
        if not self.is_active:
            print("Договор не действует!")
            return
        if obligation_index < 0 or obligation_index >= len(self.obligations):
            print("Неверный индекс обязательства!")
            return
        obligation = self.obligations[obligation_index]
        obligation["text"] += " ✓ Выполнено"
        print(f"Обязательство выполнено: {obligation['text']}")

    def terminate(self, reason):
        """Расторгнуть договор."""
        self.is_active = False
        self.status = f"расторгнут: {reason}"
        print(f"Договор №{self.contract_number} расторгнут. Причина: {reason}")

    def show_info(self):
        """Вывести полную информацию о договоре."""
        print(f"\n{'='*40}")
        print(f"  Договор №{self.contract_number}")
        print(f"{'='*40}")
        print(f"  Стороны:    {self.party_a} ↔ {self.party_b}")
        print(f"  Предмет:    {self.subject}")
        print(f"  Дата:       {self.date}")
        print(f"  Подписан:   {self.is_signed}")
        print(f"  Действует:  {self.is_active}")
        print(f"  Статус:     {self.status}")
        print(f"  Подписи:    {self.signatures}")
        print(f"  Сроки:      {len(self.deadlines)}")
        for dl in self.deadlines:
            print(f"    → {dl['description']}: {dl['date']}")
        print(f"  Обязательства: {len(self.obligations)}")
        for i, ob in enumerate(self.obligations):
            print(f"    {i}. [{ob['party']}] {ob['text']}")
        print(f"{'='*40}\n")


# === Пример использования ===
if __name__ == "__main__":
    contract = Contract(
        contract_number="ДОГ-2025/001",
        party_a="Иванов И.И.",
        party_b="ООО СтройМонтаж",
        subject="Оказание строительных услуг",
        date="15.01.2025"
    )

    contract.sign("Иванов И.И.")
    contract.sign("Генеральный директор ООО СтройМонтаж Сидоров А.В.")

    contract.add_deadline("01.03.2025", "Сдача первого этапа")
    contract.add_deadline("15.05.2025", "Полная сдача объекта")

    contract.add_obligation("ООО 'СтройМонтаж'", "Выполнить строительные работы")
    contract.add_obligation("Иванов И.И.", "Оплатить 500 000 руб.")

    contract.fulfill_obligation(1)

    contract.show_info()
```
